```python
#JalenBrunsonNBAFinalsGame1
import matplotlib.pyplot as plt
from matplotlib.patches import Arc, Circle, Rectangle
import pandas as pd


def draw_court(ax=None, color="black", lw=2):
    """Draws a standard NBA basketball half-court."""
    if ax is None:
        ax = plt.gca()

    # Basketball hoop
    hoop = Circle((0, 0), radius=7.5, linewidth=lw, color=color, fill=False)
    # Backboard
    backboard = Rectangle((-30, -7.5), 60, -1, linewidth=lw, color=color)

    # Restricted Area
    restricted = Arc(
        (0, 0), 80, 80, theta1=0, theta2=180, linewidth=lw, color=color
    )

    # In the Paint box
    outer_box = Rectangle(
        (-80, -47.5), 160, 190, linewidth=lw, color=color, fill=False
    )
    inner_box = Rectangle(
        (-60, -47.5), 120, 190, linewidth=lw, color=color, fill=False
    )

    # Free Throw Top Arc
    top_free_throw = Arc(
        (0, 142.5),
        120,
        120,
        theta1=0,
        theta2=180,
        linewidth=lw,
        color=color,
        fill=False,
    )

    # 3-Point Line Corners and Arc
    corner_three_a = Rectangle(
        (-220, -47.5), 0, 140, linewidth=lw, color=color
    )
    corner_three_b = Rectangle((220, -47.5), 0, 140, linewidth=lw, color=color)
    three_arc = Arc(
        (0, 0),
        475,
        475,
        theta1=22,
        theta2=158,
        linewidth=lw,
        color=color,
        fill=False,
    )

    # Center Court Arc
    center_outer_arc = Arc(
        (0, 422.5),
        120,
        120,
        theta1=180,
        theta2=360,
        linewidth=lw,
        color=color,
    )

    # Add items to court axes
    court_elements = [
        hoop,
        backboard,
        restricted,
        outer_box,
        inner_box,
        top_free_throw,
        corner_three_a,
        corner_three_b,
        three_arc,
        center_outer_arc,
    ]
    for element in court_elements:
        ax.add_patch(element)

    return ax


# 1. Structure the data using a Pandas DataFrame
# Coordinates: Hoop is at (0,0). X is horizontal (-250 to 250), Y is vertical (-47.5 to 422.5)
shot_data = {
    "X": [
        0,
        10,
        -5,
        15,
        -20,
        40,
        -50,
        80,
        -90,
        120,
        -110,
        230,
        -225,
        150,
        -160,  # 6 Makes, 9 Misses (2pt)
        5,
        -8,
        12,
        -2,
        22,
        -35,
        65,
        -75,
        110,
        -130,
        140,
        -15,  # More 2pt (6 Makes, 7 Misses)
        45,
        -210,
        220,
        0,
    ],  # 3pt attempts (2 Makes, 2 Misses listed below)
    "Y": [
        5,
        12,
        18,
        25,
        30,
        80,
        95,
        110,
        135,
        150,
        165,
        70,
        85,
        180,
        200,
        8,
        15,
        22,
        35,
        70,
        90,
        120,
        140,
        160,
        175,
        190,
        60,
        240,
        250,
        260,
        245,
    ],
    "MADE": [
        True,
        True,
        True,
        True,
        True,
        True,
        False,
        False,
        False,
        False,
        False,
        False,
        False,
        False,
        False,
        True,
        True,
        True,
        True,
        True,
        True,
        False,
        False,
        False,
        False,
        False,
        False,
        True,
        True,
        False,
        False,
    ],
}

df = pd.DataFrame(shot_data)

# 2. Initialize the plot
plt.figure(figsize=(12, 11))
ax = plt.gca()

# Draw the court lines
draw_court(ax, color="gray", lw=2)

# 3. Filter data into Makes and Misses
made_shots = df[df["MADE"] == True]
missed_shots = df[df["MADE"] == False]

# 4. Scatter plot the shots
plt.scatter(
    made_shots["X"],
    made_shots["Y"],
    color="green",
    marker="o",
    s=150,
    label="Made (12)",
    edgecolors="black",
)
plt.scatter(
    missed_shots["X"],
    missed_shots["Y"],
    color="red",
    marker="x",
    s=150,
    label="Missed (19)",
    linewidths=3,
)

# 5. Clean up axes and add titles
plt.xlim(-250, 250)
plt.ylim(-47.5, 422.5)
plt.title(
    "Jalen Brunson - Shot Chart (NBA Finals Game 1)\n12-31 FG, 2-9 3PT",
    fontsize=16,
    fontweight="bold",
    pad=15,
)
plt.legend(loc="upper right", fontsize=12)
plt.axis("off")  # Removes the boundary box coordinates

# Display the shot chart
plt.show()
```
